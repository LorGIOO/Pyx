# Presión de viento (cálculo ficticio)
Cuaderno vinculado desde el Anejo A con `\pynotebook{}`. Se puede ejecutar aquí, en Jupyter, como cualquier cuaderno.

```python
import numpy as np

v_b = 27.0          # velocidad básica (m/s)
rho_aire = 1.25     # densidad del aire (kg/m³)
z_tablero = 6.0     # altura del tablero sobre el terreno (m)

q_b = 0.5 * rho_aire * v_b**2 / 1000   # presión básica (kN/m²)
q_b
```

```python
def ce(z, k=0.19, z0=0.05, zmin=2.0):
    """Coeficiente de exposición (fórmula simplificada e inventada)."""
    z = max(z, zmin)
    return k**2 * np.log(z / z0) * (np.log(z / z0) + 7)

def presion(z):
    """Presión de viento a la altura z (kN/m²)."""
    return q_b * ce(z)

presion(z_tablero)
```

```python
def filas_presion(alturas):
    """Filas de una tabla LaTeX: altura, coeficiente y presión."""
    filas = []
    for z in alturas:
        filas.append(f"{z} & {ce(z):.2f} & {presion(z):.3f} \\\\")
    return "\n".join(filas)

print(filas_presion([2, 5, 10]))
```
