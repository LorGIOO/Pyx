```python
import numpy as np

v_b = 27.0          # basic wind speed (m/s)
rho_air = 1.25      # air density (kg/m³)
z_deck = 6.0        # height of the deck above ground (m)

q_b = 0.5 * rho_air * v_b**2 / 1000   # basic pressure (kN/m²)
q_b
```

```python
def ce(z, k=0.19, z0=0.05, zmin=2.0):
    """Exposure factor (simplified, made-up formula)."""
    z = max(z, zmin)
    return k**2 * np.log(z / z0) * (np.log(z / z0) + 7)

def pressure(z):
    """Wind pressure at height z (kN/m²)."""
    return q_b * ce(z)

pressure(z_deck)
```

```python
def pressure_rows(heights):
    """Rows of a LaTeX table: height, factor and pressure."""
    rows = []
    for z in heights:
        rows.append(f"{z} & {ce(z):.2f} & {pressure(z):.3f} \\\\")
    return "\n".join(rows)

print(pressure_rows([2, 5, 10]))
```
